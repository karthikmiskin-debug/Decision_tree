```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/tennis_anyone_1000.xlsx")
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns='Day ID',inplace=True)
```

```python
df.head()
```

```python
df.nunique()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Outlook']=le.fit_transform(df['Outlook'])
df['Temperature']=le.fit_transform(df['Temperature'])
df['Humidity']=le.fit_transform(df['Humidity'])
df['Wind']=le.fit_transform(df['Wind'])
df['Play Tennis']=le.fit_transform(df['Play Tennis'])
```

```python
df.columns
```

```python
df.head()
```

```python
### Renaming the columns
df.rename(columns={'Play Tennis':'Play Tennis'},inplace=True)
```

```python
df['Play Tennis'].value_counts()
```

```python
df.shape
```

```python
x = df.iloc[:,0:-1]
y = df.iloc[:,-1]
```

```python
### Splitting dataset Horizointally(train and test set(20%))
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
### using Decision Tree model
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt
```

```python
#Training the model
dt.fit(x_train,y_train)
```

```python
### predict the model

predict = dt.predict(x_test)
```

```python
### Evaluation of the model
from sklearn.metrics import classification_report

print(classification_report(y_test,predict))
```

>>Decision Tree with Accuracy of 100%
